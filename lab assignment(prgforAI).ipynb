```python
#01 lab work
name = input("Enter Student Name: ")
age = int(input("Enter Student Age: "))
courses = int(input("Enter Number of Courses: "))

total = 0
for i in range(1, courses + 1):
    marks = int(input("Enter Marks in Course " + str(i) + ": "))
    total = total + marks

average = total / courses
age_after_5 = age + 5

bonus_total = total + 5 * courses
bonus_average = bonus_total / courses

print("\n----- Challenge Task Result -----")
print("Student Name:", name)
print("Total Marks:", total)
print("Average Marks:", round(average, 2))
print("Age After 5 Years:", age_after_5)
print("Average Marks After 5 Bonus Marks:", round(bonus_average, 2))
```

```python
#02 lab work
name = input("Enter student name: ")
m1 = int(input("Enter marks in Subject 1: "))
m2 = int(input("Enter marks in Subject 2: "))
m3 = int(input("Enter marks in Subject 3: "))

total = m1 + m2 + m3
percentage = total / 300 * 100

if percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "F"

if percentage >= 50:
    status = "Pass"
else:
    status = "Fail"

print("\n===== Student Report =====")
print("Name:", name)
print("Total Marks:", total, "out of 300")
print("Percentage:", round(percentage, 2), "%")
print("Grade:", grade)
print("Status:", status)
```

```python
#03 lab work
subjects = ["Programming", "Mathematics", "Artificial Intelligence"]

student_count = 0
grand_total = 0

name = input("Enter student name (-1 to stop): ")

while name != "-1":
    student_total = 0

    for i in range(3):
        subject = subjects[i]
        marks = int(input("Enter " + subject + " marks: "))
        student_total += marks

    average = student_total / 3
    student_count += 1
    grand_total += student_total

    print("\nStudent:", name)
    print("Total Marks:", student_total)
    print("Average Marks:", round(average, 2))

    print("Multiplication table (1 to 5) for", name)
    for row in range(1, 6):
        for col in range(1, 6):
            print(str(row * col).ljust(4), end="")
        print()
    print()

    name = input("Enter student name (-1 to stop): ")

print("\n----- Final Result -----")
print("Total Students:", student_count)
if student_count > 0:
    overall_average = grand_total / (student_count * 3)
    print("Overall Average:", round(overall_average, 2))
else:
    print("No student was entered")
```

```python
#04th lab work
PASS_MARKS = 50

def calculate_total(m1, m2, m3):
    total = m1 + m2 + m3
    return total

def calculate_percentage(total):
    percentage = total / 300 * 100
    return percentage

def find_grade(percentage):
    if percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"

def display_result(name, total, percentage, grade):
    if percentage >= PASS_MARKS:
        status = "Pass"
    else:
        status = "Fail"
    print("\n===== Student Result =====")
    print("Name:", name)
    print("Total Marks:", total)
    print("Percentage:", round(percentage, 2), "%")
    print("Grade:", grade)
    print("Status:", status)

def highest_subject(subject_names, marks_list):
    highest = max(marks_list)
    position = marks_list.index(highest)
    print("Highest Scoring Subject:", subject_names[position], "-", highest, "marks")

name = input("Enter student name: ")
m1 = int(input("Enter marks in Programming: "))
m2 = int(input("Enter marks in Mathematics: "))
m3 = int(input("Enter marks in AI: "))

total = calculate_total(m1, m2, m3)
percentage = calculate_percentage(total)
grade = find_grade(percentage)

display_result(name, total, percentage, grade)
highest_subject(["Programming", "Mathematics", "AI"], [m1, m2, m3])
```

```python
#05th lab work
def calculate_average(m1, m2, m3):
    return (m1 + m2 + m3) / 3

def is_passed(average):
    if average >= 50:
        return True
    else:
        return False

def get_grade(average):
    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"

def student_report(name, m1, m2, m3):
    average = calculate_average(m1, m2, m3)
    grade = get_grade(average)
    passed = is_passed(average)
    return average, grade, passed

name = input("Enter student name: ")
m1 = int(input("Enter marks in subject 1: "))
m2 = int(input("Enter marks in subject 2: "))
m3 = int(input("Enter marks in subject 3: "))

average, grade, passed = student_report(name, m1, m2, m3)

print("\n===== Student Report =====")
print("Name:", name)
print("Marks:", m1, m2, m3)
print("Average:", round(average, 2))
print("Grade:", grade)
print("Passed:", passed)
```
